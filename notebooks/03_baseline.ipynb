# 03 — Interpolation baseline

What reconstruction MSE does a non-learned method get? Run the **energy-shell angular-mean** baseline on a small validation set so you have a number to beat.


```python
import sys, os, numpy as np
ROOT = os.path.dirname(os.path.abspath('.')) if os.path.basename(os.getcwd())=='notebooks' else os.path.abspath('.')
SRC = os.path.join(ROOT, 'src')
if SRC not in sys.path: sys.path.insert(0, SRC)
import data_pipeline as dp
from baseline import evaluate_baselines
```

```python
files = dp.find_dist_files(os.path.join(ROOT, 'MMS-FPI-Data-Gaps'))
# pick the held-out validation files using the same split the trainer does
_, val = dp.split_files(files, val_fraction=0.2, seed=0)
print('val files:', len(val))
cubes = np.concatenate([dp.read_dist_file(f, subsample=24).cubes for f in val], axis=0)
print('val cubes:', cubes.shape)
```

```python
mask = dp.synthetic_wedge_mask()
result = evaluate_baselines(cubes, mask)
for name, m in result.items():
    print(f'{name:25s}  masked MSE = {m["masked_mse"]:.4e}   masked MAE = {m["masked_mae"]:.4e}')
```
