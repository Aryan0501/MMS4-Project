# 04 — Train from scratch

Wraps `src/train_incremental.py`. Edit the config in the next cell, then run. Outputs land in `outputs/<RUN_NAME>/`.

**Models** (`MODEL`):
- `unet` — 3-D U-Net (recommended). Add `--temporal-window 1 --features pitch_angle,logb` for the headline config.
- `convlstm` — ConvLSTM "cube through time".
- `jepa` — I-JEPA-lite (predict embeddings in latent space; no decoder).

**Time on CPU**: at the defaults below, ~60-90 s per file × 21 train files × `--rounds`.


```python
import os
ROOT = os.path.dirname(os.path.abspath('.')) if os.path.basename(os.getcwd())=='notebooks' else os.path.abspath('.')
SRC = os.path.join(ROOT, 'src'); DATA = os.path.join(ROOT, 'MMS-FPI-Data-Gaps')

RUN_NAME   = 'unet_from_notebook'
MODEL      = 'unet'              # 'unet' | 'convlstm' | 'jepa'
EXTRA      = '--temporal-window 1 --features pitch_angle,logb --dump-embeddings'
ROUNDS     = 1                    # bump to 2-3 for a longer learning curve
INNER_EPS  = 2
SUBSAMPLE  = 20                   # CPU speed knob: keep every Nth burst timestep
BASE_F     = 10
BATCH_SIZE = 64
MASK       = 'wedge'              # 'wedge' (synthetic) | 'explosion' (realistic)
OUT        = os.path.join(ROOT, 'outputs', RUN_NAME)
print('out:', OUT)
```

```python
cmd = (f'python {SRC}/train_incremental.py --data-root {DATA} --out {OUT} '
       f'--model {MODEL} {EXTRA} --mask {MASK} --rounds {ROUNDS} '
       f'--inner-epochs {INNER_EPS} --subsample {SUBSAMPLE} --base-filters {BASE_F} '
       f'--batch-size {BATCH_SIZE}')
print(cmd)
```
