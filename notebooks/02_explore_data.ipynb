# 02 — Explore one CDF file

Load a `des-dist` CDF, look at the raw distribution, the model-space transform, and the occlusion mask.


```python
import sys, os, numpy as np, matplotlib.pyplot as plt
ROOT = os.path.dirname(os.path.abspath('.')) if os.path.basename(os.getcwd())=='notebooks' else os.path.abspath('.')
SRC = os.path.join(ROOT, 'src')
if SRC not in sys.path: sys.path.insert(0, SRC)
import data_pipeline as dp
```

```python
files = dp.find_dist_files(os.path.join(ROOT, 'MMS-FPI-Data-Gaps'))
print(len(files), 'des-dist files found')
fb = dp.read_dist_file(files[0], subsample=8)
print('shape', fb.cubes.shape, 'min', fb.cubes.min(), 'max', fb.cubes.max())
```

## One distribution at one timestep — energy spectrum and angular slice


```python
cube = fb.cubes[0]                               # (32 energy, 16 theta, 32 phi) in MODEL SPACE [0,1]
phys = dp.to_physical_space(cube)                # back to PSD (s^3/cm^6)
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(phys.sum(axis=(1,2)))
ax[0].set_xlabel('energy bin'); ax[0].set_ylabel('integrated PSD'); ax[0].set_yscale('log'); ax[0].set_title('energy spectrum (one timestep)')
im = ax[1].imshow(cube[16], aspect='auto', cmap='viridis')
ax[1].set_xlabel('phi bin'); ax[1].set_ylabel('theta bin'); ax[1].set_title('one energy slice (model space)'); plt.colorbar(im, ax=ax[1])
plt.tight_layout(); plt.show()
```

## The occlusion masks


```python
wedge = dp.synthetic_wedge_mask()
explosion = dp.load_explosion_mask(ROOT)
print('wedge: occluded fraction', wedge.mean(), '   explosion: occluded fraction', explosion.mean())
fig, ax = plt.subplots(1, 2, figsize=(9, 4))
ax[0].imshow(wedge[16], cmap='Greys'); ax[0].set_title('synthetic wedge (one E shell)')
ax[1].imshow(explosion[16], cmap='Greys'); ax[1].set_title('explosion.npy (one E shell)')
plt.tight_layout(); plt.show()
```
