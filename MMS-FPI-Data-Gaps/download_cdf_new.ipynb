```python
import requests
from bs4 import BeautifulSoup

# URL of the page with FGM files for March 2024
#url = 'https://lasp.colorado.edu/mms/sdc/public/about/browse/mms1/fpi/brst/l2/des-moms/2024/03/15/'
#url = 'https://lasp.colorado.edu/mms/sdc/public/about/browse/mms1/fpi/brst/l2/des-dist/2024/03/15/'
url = 'https://lasp.colorado.edu/mms/sdc/public/about/browse/mms1/fgm/brst/l2/2024/03/15/'


# Send a GET request to the webpage
response = requests.get(url)
if response.status_code == 200:
    # Parse the HTML content
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Find all <a> tags that have links to CDF files
    links = []
    for link in soup.find_all('a'):
        href = link.get('href')
        if href and href.endswith('.cdf'):
            full_link = url + href
            links.append(full_link)
            print(full_link)
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")

# Save links to a text file if needed
with open('fgm_links_march.txt', 'w') as file:
    for link in links:
        file.write(link + '\n')
```

```python
import os
import requests

# Load links from the text file
with open('fgm_links_march.txt', 'r') as file:
    links = file.readlines()

# Create a folder to store the downloaded files
output_folder = 'data_03_15/fgm'
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Download each file
for link in links:
    link = link.strip()  # Remove any trailing newlines or spaces
    filename = os.path.join(output_folder, link.split('/')[-1])
    
    print(f"Downloading {filename}...")
    response = requests.get(link)
    if response.status_code == 200:
        with open(filename, 'wb') as file:
            file.write(response.content)
        print(f"Successfully downloaded {filename}")
    else:
        print(f"Failed to download {filename}. Status code: {response.status_code}")
```
