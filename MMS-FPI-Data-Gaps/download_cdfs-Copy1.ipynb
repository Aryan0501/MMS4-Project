```python
file_paths = []

for url in urls:
    # Split the URL into components
    parts = url.split('/')
    # Find the index of 'browse' to remove the base URL components
    try:
        browse_index = parts.index('browse')
    except ValueError:
        print(f"'browse' not found in URL: {url}")
        continue
    # Extract the path components after 'browse'
    path_components = parts[browse_index + 1:]  # Skip 'browse' itself
    # Prepend 'data' to the path
    local_path = os.path.join('data', *path_components)
    file_paths.append(local_path)

# Now file_paths contains the desired local file paths
for path in file_paths:
    print(path)
```

```python
def download_file(url, file_path):
    print(f"Downloading: {url}")
    # Create directories if they don't exist
    os.makedirs(os.path.dirname(file_path), exist_ok=True)
    # Download the file
    response = requests.get(url)
    response.raise_for_status()  # Check if the download was successful

    # Write the file to disk
    with open(file_path, 'wb') as file:
        file.write(response.content)


# Download all files
for url, file_path in zip(urls, file_paths):
    download_file(url, file_path)
```
