```python
# Specify the path to your CDF file
import cdflib

# Specify the path to your CDF file
cdf_file_path = 'mms1_fpi_fast_l2_dis-momsaux_20240615180000_v3.4.0.cdf'

# Open the CDF file
cdf_file = cdflib.CDF(cdf_file_path)


```

```python
# Open the CDF file
cdf_file = cdflib.CDF(cdf_file_path)

# Get the list of variables
variables = cdf_file.cdf_info().zVariables

# Initialize an empty dictionary to hold data
data_dict = {}

# Loop through each variable and add its data to the dictionary
for var in variables:
    try:
        data = cdf_file.varget(var)
        # Flatten the data if it's multidimensional and convert to list
        if data.ndim > 1:
            data = data.flatten()
        data_dict[var] = data
    except Exception as e:
        print(f"Could not load data for variable {var}: {e}")

```
